# CliniScan+ 01: Dataset Analysis & Exploratory Data Analysis (EDA)
This notebook inspects the VinBigData Chest X-ray dataset, evaluates class distributions, analyzes bounding box aspect ratios, and studies co-occurrence patterns across the 10 target lung abnormality classes.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

DATA_DIR = r'../vinbigdata'
train_csv = os.path.join(DATA_DIR, 'train.csv')
df = pd.read_csv(train_csv)
print(f'Total records: {len(df):,}')
print(f'Unique images: {df["image_id"].nunique():,}')
df.head()
```

```python
# 10 Selected Lung Target Classes
TARGET_CLASSES = [
    'Atelectasis', 'ILD', 'Pulmonary fibrosis', 'Pneumothorax',
    'Pleural effusion', 'Pleural thickening', 'Consolidation',
    'Lung Opacity', 'Nodule/Mass', 'Infiltration'
]

target_df = df[df['class_name'].isin(TARGET_CLASSES)].copy()
print(f'Target abnormality annotations: {len(target_df):,}')
print(f'Positive CXR images: {target_df["image_id"].nunique():,}')

plt.figure(figsize=(10, 5))
sns.countplot(data=target_df, y='class_name', order=target_df['class_name'].value_counts().index, palette='viridis')
plt.title('Distribution of 10 Selected Lung Abnormalities in VinBigData')
plt.xlabel('Annotation Count')
plt.ylabel('Abnormality Class')
plt.tight_layout()
plt.show()
```

```python
# Bounding Box Geometry Analysis
target_df['bbox_w'] = target_df['x_max'] - target_df['x_min']
target_df['bbox_h'] = target_df['y_max'] - target_df['y_min']
target_df['bbox_area_norm'] = (target_df['bbox_w'] * target_df['bbox_h']) / (target_df['width'] * target_df['height'])
target_df['aspect_ratio'] = target_df['bbox_w'] / (target_df['bbox_h'] + 1e-6)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.boxplot(data=target_df, x='aspect_ratio', y='class_name', ax=axes[0], palette='magma')
axes[0].set_title('Aspect Ratio (Width/Height) by Class')
axes[0].set_xlim(0, 4)

sns.boxplot(data=target_df, x='bbox_area_norm', y='class_name', ax=axes[1], palette='crest')
axes[1].set_title('Normalized Bounding Box Area by Class')
axes[1].set_xlim(0, 0.5)
plt.tight_layout()
plt.show()
```
