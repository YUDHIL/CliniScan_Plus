# CliniScan+ 03: YOLO11 Model Evaluation & Error Analysis
Executes model validation, plots Precision-Recall curves, confusion matrices, and sweeps confidence thresholds for clinical deployment.

```python
from ultralytics import YOLO
import os

model_path = r'../models/cliniscan_yolo11n_best.pt'
if not os.path.exists(model_path):
    model_path = 'yolo11n.pt'

model = YOLO(model_path)
metrics = model.val(data=r'../dataset.yaml', split='val')
print(f'Overall mAP@50: {metrics.box.map50*100:.2f}%')
print(f'Overall mAP@50-95: {metrics.box.map*100:.2f}%')
print(f'Precision: {metrics.box.mp*100:.2f}%')
print(f'Recall: {metrics.box.mr*100:.2f}%')
```
